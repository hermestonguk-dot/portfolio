```python
import pandas as pd
from pathlib import Path

try:
    input_csv = snakemake.input[0]
    iso3 = snakemake.wildcards.iso3
    outputs = snakemake.output
except NameError:
    iso3 = "GBR"
    input_csv = "../../results/only_countries.csv"
    outputs = {
        "total": "../../results/by_country/GBR/total.csv",
        "coal": "../../results/by_country/GBR/coal.csv",
        "oil": "../../results/by_country/GBR/oil.csv",
        "gas": "../../results/by_country/GBR/gas.csv",
        "cement": "../../results/by_country/GBR/cement.csv",
        "flaring": "../../results/by_country/GBR/flaring.csv",
        "other": "../../results/by_country/GBR/other.csv",
    }
```

```python

df = pd.read_csv(input_csv)

df_country = df[df["iso3166"] == iso3].copy()
df_country = df_country.sort_values("Year")
```

```python
DIMENSIONS = [
    ("Total",   "total"),
    ("Coal",    "coal"),
    ("Oil",     "oil"),
    ("Gas",     "gas"),
    ("Cement",  "cement"),
    ("Flaring", "flaring"),
    ("Other",   "other"),
]

for col_name, key in DIMENSIONS:
    sub = df_country[["Year", col_name]].copy()

    sub = sub.dropna(subset=[col_name])

    out_path = outputs[key]
    sub.to_csv(out_path, index=False)

```
