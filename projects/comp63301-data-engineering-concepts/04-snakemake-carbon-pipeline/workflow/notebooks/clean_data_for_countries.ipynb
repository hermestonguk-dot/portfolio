```python
import pandas as pd
from pathlib import Path

try:
    input_csv = snakemake.input[0]
    only_countries_csv = snakemake.output[0]
    countries_csv = snakemake.output[1]
except NameError:
    input_csv = "../results/validated.csv"
    only_countries_csv = "../results/only_countries.csv"
    countries_csv = "../results/countries.csv"


df = pd.read_csv(input_csv)


df = df.rename(
    columns={
        "ISO 3166-1 alpha-3": "iso3166",
        "UN M49": "un_m49",
        "Per Capita": "per_capita",
    }
)


df_countries = df[df["iso3166"].notna()].copy()


non_countries = [
    "Global",
    "International Aviation",
    "International Shipping",
    "Antarctica",
]
df_countries = df_countries[~df_countries["Country"].isin(non_countries)].copy()


countries_df = df_countries.copy()


countries = df_countries[["Country", "iso3166", "un_m49"]].drop_duplicates()


for p in [only_countries_csv, countries_csv]:
    Path(p).parent.mkdir(parents=True, exist_ok=True)


countries_df.to_csv(only_countries_csv, index=False)
countries.to_csv(countries_csv, index=False)
```
